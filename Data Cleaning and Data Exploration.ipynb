```python
df_1['Customer ID'] =df_1['Customer ID'].astype('Int64')
```

```python
df_2 = pd.read_excel('online_retail_II.xlsx',sheet_name='Year 2010-2011')
```

```python
df_2.head()
```

```python

# change datatype - float to int 
df_2['Customer ID'] =df_2['Customer ID'].astype('Int64')
```

```python
print(df_1.shape)

print(df_2.shape)
```

```python
# Before appending , check if columns matches

print(df_1.columns.tolist())
print(df_2.columns.tolist())
print(df_1.columns.equals(df_2.columns))  
```

```python
# Combine into one dataframe
df = pd.concat([df_1, df_2], ignore_index=True)

print("Original Shape :",df.shape)
print("Original Columns :",df.columns)
```

```python
print(df.head())
print(df.info())
```

```python
df.isnull().sum()
```

```python
# Remove cancelled order , ( Invoice record starting with 'C' means cancelled )
df=df[~df['Invoice'].astype(str).str.startswith('C')]
```

```python
# Removing null values : Description col
df=df.dropna(subset=['Description'])

df.isnull().sum()
```

```python
# Keep only positive Quantity and Price  
df=df[(df['Quantity'] > 0) & (df['Price'] > 0)]
```

```python
# Create Revenue  col 
df['Revenue'] = df['Quantity'] * df['Price']
```

```python
df.head()
print("Cleaned Shape :",df.shape)
```

```python
# Saving File 
df.to_csv('cleaned_retail.csv',index='False')
```
